### Data dictionary

| Group | Columns | Description |
|---|---|---|
| Identifier | `customerID` | Unique customer ID (no predictive value) |
| Demographics | `gender`, `SeniorCitizen`, `Partner`, `Dependents` | Who the customer is |
| Account | `tenure`, `Contract`, `PaperlessBilling`, `PaymentMethod`, `MonthlyCharges`, `TotalCharges` | Relationship length and billing |
| Phone services | `PhoneService`, `MultipleLines` | Phone products |
| Internet services | `InternetService`, `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies` | Internet products and add-ons |
| **Target** | `Churn` | Customer left in the last month |

Two things stand out immediately from `.info()`:
- `TotalCharges` should be numeric but was read as text.
- `SeniorCitizen` is encoded as `0/1` while every other binary column uses `Yes/No`.
